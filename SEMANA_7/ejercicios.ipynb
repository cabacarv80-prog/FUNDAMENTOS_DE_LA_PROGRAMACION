```python
import time
import sys

def barra_de_carga(total, longitud=100):
    for i in range(total + 1):
        porcentaje = (i / total) * 100
        lleno = int(longitud * i // total)
        # Dibujar los bloques llenos (█) y vacíos (-)
        barra = '█' * lleno + ' ' * (longitud - lleno)
        
        # \r devuelve el cursor al inicio de la misma línea
        sys.stdout.write(f'\rCargando: [{barra}] {porcentaje:.0f} %')
        sys.stdout.flush()
        
        # Simula que Python está haciendo un trabajo pesado
        time.sleep(0.03)
    
    print("\n¡Completado con éxito!")

# Ejecutar con 100 pasos
if __name__ == "__main__":
    barra_de_carga(100)
```

```python
# Inicialización de variables del sistema (Valores iniciales)
stock_ml = 5000         # Inventario inicial en mililitros (ejemplo: 5 litros)
total_ventas = 0.0      # Acumulador de dinero recaudado
clientes = 0            # Contador de clientes atendidos

opcion = -1             # Variable de control del menú

# Ciclo principal del programa
while opcion != 4:
    print("\n--- MENÚ DE OPCIONES ---")
    print("1. Realizar una venta")
    print("2. Consultar stock disponible")
    print("3. Mostrar resumen de cierre")
    print("4. Salir del programa")
    
    try:
        opcion = int(input("Seleccione una opción (1-4): "))
    except ValueError:
        print("Por favor, ingrese un número válido.")
        continue

    # Opción 1: Proceso de Venta
    if opcion == 1:
        print("\n--- PRODUCTOS DISPONIBLES ---")
        print("1. Producto 1 ($35 - 100 ml)")
        print("2. Producto 2 ($60 - 200 ml)")
        print("3. Producto 3 ($220 - 1000 ml)")
        
        try:
            producto = int(input("Seleccione el producto (1-3): "))
            cantidad = int(input("Ingrese la cantidad a comprar: "))
        except ValueError:
            print("Entrada inválida. Intente de nuevo.")
            continue
            
        precio = 0
        ml = 0
        
        # Asignación de precio y ml según el producto
        if producto == 1:
            precio = 35
            ml = 100
        elif producto == 2:
            precio = 60
            ml = 200
        elif producto == 3:
            precio = 220
            ml = 1000
        else:
            print("ERROR: Producto no válido.")
            continue # Regresa al menú principal
            
        # Validación de Stock y cálculo de totales
        if (ml * cantidad) <= stock_ml:
            subtotal = precio * cantidad
            
            if subtotal > 300:
                descuento = subtotal * 0.10
            else:
                descuento = 0
                
            total_pagar = subtotal - descuento
            stock_ml = stock_ml - (ml * cantidad)
            total_ventas = total_ventas + total_pagar
            clientes = clientes + 1
            
            print(f"\n¡Venta exitosa!")
            print(f"Subtotal: ${subtotal:.2f}")
            print(f"Descuento aplicado: ${descuento:.2f}")
            print(f"Total a pagar: ${total_pagar:.2f}")
        else:
            print("\nERROR: Stock insuficiente.")
            
    # Opción 2: Consultar Stock
    elif opcion == 2:
        print(f"\nStock disponible: {stock_ml} ml")
        
    # Opción 3: Resumen de Cierre
    elif opcion == 3:
        print("\n=== RESUMEN DE CIERRE ===")
        print(f"Clientes atendidos: {clientes}")
        print(f"Ventas Totales: ${total_ventas:.2f}")
        
    # Opción 4: Salida
    elif opcion == 4:
        print("\nGracias por usar el sistema. ¡Hasta luego!")
    else:
        print("Opción no válida. Intente de nuevo.")
```

```python
USUARIOS_AUTORIZADOS = {
    "administrador": "Administrador de Inventario",
    "gerente": "Gerente general"
}

INTENTOS_MAXIMOS = 5
acceso_concedido = False
usuario_activo = None


def iniciar_sesion():
    global acceso_concedido, usuario_activo
    
    intentos_restantes = INTENTOS_MAXIMOS
    
    print("==================================================")
    print("   SISTEMA DE CONTROL DE INVENTARIO - HELADERÍA   ")
    print("==================================================")
    
    while not acceso_concedido and intentos_restantes > 0:
        entrada = input("\nIngrese su usuario (Gerente / Administrador): ").strip().lower()
        
        if entrada in USUARIOS_AUTORIZADOS:
            acceso_concedido = True
            usuario_activo = USUARIOS_AUTORIZADOS[entrada]
            print(f"\n[OK] Autenticación exitosa. ¡Bienvenido/a, {usuario_activo}!")
        else:
            intentos_restantes -= 1
            if intentos_restantes > 0:
                print(f"[ERROR] Usuario no autorizado. Intentos restantes: {intentos_restantes}")
            else:
                print("\n[ALERTA] Has alcanzado el número máximo de intentos fallidos.")
                print("[SISTEMA] Acceso denegado. Cerrando aplicación...")


    return usuario_activo


usuario_sesion = iniciar_sesion()

if acceso_concedido:
    print(f"\n---> Sesión iniciada correctamente como: '{usuario_sesion}'")
```

```python
import sys

# Matriz 2x2 de opciones (filas x columnas)
menu_matriz = [
    ["[1] Perfil",   "[2] Configuración"],
    ["[3] Informes", "[4] Salir"]
]

# Bucle principal que mantiene el menú activo
while True:
    print("\n" + "=" * 40)
    print("        MENÚ PRINCIPAL (MATRIZ)")
    print("=" * 40)
    
    # Recorrido e impresión de la matriz fila por fila
    for fila in menu_matriz:
        # Imprime los elementos de la fila alineados en columnas
        print(f"{fila[0]:<20} {fila[1]:<20}")
    
    print("=" * 40)
    opcion = input("Selecciona una opción (1-4): ").strip()
    
    # Control de decisiones
    if opcion == "1":
        print("\n-> Has entrado al Perfil de Usuario.")
    elif opcion == "2":
        print("\n-> Has entrado a Configuración del Sistema.")
    elif opcion == "3":
        print("\n-> Generando Informes...")
    elif opcion == "4":
        print("\n-> Saliendo del programa...")
        sys.exit(0)  # Finaliza la ejecución usando el módulo sys
    else:
        print("\n[!] Opción no válida. Inténtalo de nuevo.")
```
