# Extra 1: División de cuenta con propina

```python
c = float(input("Cual es el total de la cuenta?"))
p = float(input("Cual es el porcentaje de propina?"))
s = float(input("En cuantas personas se va a dividir la cuenta?"))

propina = (c * p / 100)
total = c + propina

print(f"La propina es: {propina}")
print(f"El total es: {total}")
print(f"Cada persona debe pagar: {total / s}")
```

# Extra 2: Conversor de minutos a días, horas y minutos

```python
m = int(input("Minutos: "))

d = m//1440
mr =m%1440 

h= mr//60
m_rest = mr%60

print(f"{d} días, {h} horas y {m_rest} minutos")
```

# Extra 3: Calificación final ponderada

```python
print("SUMA DE VALORES")

parcial1= int(input("Ingrese el primer valor: "))
parcial2= int(input("Ingrese el segundo valor: "))
parcial3= int(input("Ingrese el tercer valor: "))

suma = parcial1*0.3 + parcial2*0.3 + parcial3*0.4
print("La suma de los valores es:", suma)
```
