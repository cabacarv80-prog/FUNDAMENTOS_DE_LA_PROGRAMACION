# CALCULADORA DE TIEMPO EN DIGITAL

```python
nombre_del_estudiante=input("nombre del estudiante:")
zelda=float(input("ingresa cuanto tiempo le dedicas a zelda: "))
instagram=float(input("ingresa cuanto tiempo le dedicas a instagram: "))
tiktok=float(input("ingresa cuanto tiempo le dedicas a tiktok: "))
mario_kart=float(input("ingresa cuanto tiempo le dedicas a mario kart: "))
facebook=float(input("ingresa cuanto tiempo le dedicas a facebook: "))
vicios_totales=(zelda+instagram+tiktok+mario_kart+facebook)
porcentajes=(vicios_totales/24*100)
print(f"el estudiante {nombre_del_estudiante} tiene vicios totales de {vicios_totales} horas")
print(f"esto representa el {porcentajes:.2f}% del día")
```
