```python
def promedio (sumatoria, n):
    return sumatoria / n

n = int(input("cuantas calif son: "))
sumatoria = 0

print(f"se van a pedir {n} calificaciones")
for c in range(1, n+1):
    calif = float(input(f"dame la calificación{c}"))
    sumatoria = sumatoria + calif

print(f"el promedio de {n} las calificaciones es: {promedio(sumatoria, n)}")
```

```python
#Crea una variable de texto llamada "palabra' y le asigna el valor "Hola"
palabra = "hello kitty"

#Recorre cada letra de "palabra', convierte cada una a su número ASCII con ord() y guarda los números en una lista 
valores_ascii = [ord (caracter) for caracter in palabra]

#Comentario explicativo del paso 1 (Python ignora las lineas que empiezan con #)
#1. Tomamos cada número y le sumamos 3 para codificarlo

#Crea una nueva lista donde a cada número de la lista 'valores asci se le suman 3 unidades 
valores_codificados = [numero +3 for numero in valores_ascii]

#Comentario explicativo del paso 2
#2. Convertimos los nuevos números de vuelta a letras usando chr()

#Recorre la lista de números alterados y usa chr() para transformar cada número de vuelta a su nueva letra 
letras_codificadas = [chr(numero) for numero in valores_codificados]

#Une todas las letras individuales de la lista "letras codificadas" en un solo texto continuo stilizando un separador vacío""

palabra_encriptada = "".join(letras_codificadas)

#Muestra en pantalla la lista de números ASCII originales y la palabra original separando con commas

print("Original", valores_ascii, "->", palabra)

#Muestra en pantalla la lista de números modificados y la nueva palabra encriptada separando con comas 
print("Codificado (+3):", valores_codificados, "->", palabra_encriptada)
```
