```python
calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)

print("la tercera clificación es: ", calificaciones[2])

calif5 = float(input("introduce tu numero: "))
calif6 = float(input("pon tu numero bro: "))

nuevas_calificaciones = calificaciones + (calif5, calif6)

lista_calif = list(nuevas_calificaciones)

lista_calif.sort(reverse=True)
print("mayor a menor: ", lista_calif)

def calcular_suma(elementos):
    return sum(elementos)

print("nueva tupla", nuevas_calificaciones)
print("resultado: ", calcular_suma(lista_calif))
```

## EXTRA 2 CASTIGO

```python
candidatos = {
    "A":0, 
    "B":0, 
    "C":0
    }

cantidad = int(input("¿cuantos votos van a ser?"))
conteo = 1

while conteo <= cantidad:
    voto = input(f"ingresa el voto {conteo} (A, B o C): ")
    if voto.upper() in candidatos:
        print(f"voto es valido para {voto.upper()}")
        candidatos[voto.upper()]+= 1
        conteo += 1
    else:
        print("el voto no es valido, otra vez")

print("\nResultados: ")
for k, v in candidatos.items():
    print(f"opción {k}, votos:{v} ")

print(f"\nNumero de votos: {cantidad}")
```

```python
def op1(): print(f"Ordenada: {tuple(sorted((45, 12, 89, 3, 27)))}")

def op2():
    c = {"ana": "555-1234", "carlos": "555-5678", "sofía": "555-9012"}
    n = input("Nombre: ").capitalize()
    print(f"teléfono: {c[n]}" if n in c else "No encontrado.")

def op3():
    try:
        print(f"Resultado: {float(input('Num 1: ')) / float(input('Num 2: '))}")
    except ZeroDivisionError: print("Error: División entre cero.")
    except ValueError: print("Error: Entrada no válida.")

def op4():
    t = input("Texto: ")
    print(f"Caracteres: {len(t)} | Palabras: {len(t.split())}")

while True:
    print("\n1.Tupla | 2.Contacto | 3.Dividir | 4.Texto | 5.Salir")
    op = input("Opción: ")
    if op == "1": op1()
    elif op == "2": op2()
    elif op == "3": op3()
    elif op == "4": op4()
    elif op == "5": print("Adiós."); break
    else: print("cintas error.")
```
